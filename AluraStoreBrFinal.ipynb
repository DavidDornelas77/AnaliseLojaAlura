### Importação dos dados



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

loja = pd.read_csv(url)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

loja.head()
```

#1. Análise do faturamento


```python
faturamento_loja1 = loja['Preço'].sum()
faturamento_loja2 = loja2['Preço'].sum()
faturamento_loja3 = loja3['Preço'].sum()
faturamento_loja4 = loja4['Preço'].sum()

print(f"Faturamento Loja 1: R${faturamento_loja1:.2f}")
print(f"Faturamento Loja 2: R${faturamento_loja2:.2f}")
print(f"Faturamento Loja 3: R${faturamento_loja3:.2f}")
print(f"Faturamento Loja 4: R${faturamento_loja4:.2f}")
```

```python
import matplotlib.pyplot as plt

faturamento_lojas = [faturamento_loja1, faturamento_loja2, faturamento_loja3, faturamento_loja4]
nomes_lojas = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']

# Criando o gráfico de barras horizontais
plt.figure(figsize=(8, 5))
plt.barh(nomes_lojas, faturamento_lojas)
plt.title('Faturamento por Loja')
plt.xlabel('Faturamento (R$)')
plt.ylabel('Loja')  # Rótulo do eixo y
#plt.xlim(0, 5)  # Define o limite do eixo x (ajuste conforme necessário) - removido para evitar cortes
plt.tight_layout()
plt.show()
```

# 5. Frete Médio por Loja


```python
# Cálculo da média do frete para cada loja
media_frete_loja1 = loja['Frete'].mean()
media_frete_loja2 = loja2['Frete'].mean()
media_frete_loja3 = loja3['Frete'].mean()
media_frete_loja4 = loja4['Frete'].mean()

print(f"Loja 1 - Frete médio: R${media_frete_loja1:.2f}")
print(f"Loja 2 - Frete médio: R${media_frete_loja2:.2f}")
print(f"Loja 3 - Frete médio: R${media_frete_loja3:.2f}")
print(f"Loja 4 - Frete médio: R${media_frete_loja4:.2f}")
```

```python
import matplotlib.pyplot as plt

# Média do frete por loja
media_frete_lojas = [media_frete_loja1, media_frete_loja2, media_frete_loja3, media_frete_loja4]
nomes_lojas = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']

# Criando o gráfico de linha
plt.figure(figsize=(8, 5))  # Define o tamanho da figura
plt.plot(nomes_lojas, media_frete_lojas, marker='o', linestyle='-')  # marker e linestyle para personalizar a linha
plt.title('Média do Frete por Loja')
plt.xlabel('Loja')
plt.ylabel('Média do Frete (R$)')
#plt.ylim(0, 5)  # Define o limite do eixo y (ajuste conforme necessário) - removido para evitar cortes.
plt.tight_layout()  # Ajusta o layout para evitar sobreposição de elementos
plt.grid(True) # Adiciona grade ao gráfico
plt.show()
```

```python
# Cálculo do frete total para cada loja
frete_total_loja1 = loja['Frete'].sum()
frete_total_loja2 = loja2['Frete'].sum()
frete_total_loja3 = loja3['Frete'].sum()
frete_total_loja4 = loja4['Frete'].sum()

print(f"Loja 1 - Frete total: R${frete_total_loja1:.2f}")
print(f"Loja 2 - Frete total: R${frete_total_loja2:.2f}")
print(f"Loja 3 - Frete total: R${frete_total_loja3:.2f}")
print(f"Loja 4 - Frete total: R${frete_total_loja4:.2f}")
```

# 6. Análise dos Dados disponíveis

```python
faturamento_lojas = [faturamento_loja1, faturamento_loja2, faturamento_loja3, faturamento_loja4]
total_produtos_lojas = [total_produtos_loja1, total_produtos_loja2, total_produtos_loja3, total_produtos_loja4]
media_frete_lojas = [media_frete_loja1, media_frete_loja2, media_frete_loja3, media_frete_loja4]
nomes_lojas = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']

# Calculando o faturamento, excluindo as despesas de cada loja
faturamento_menos_despesas = []
for i in range(len(nomes_lojas)):
    lucro = faturamento_lojas[i] - (total_produtos_lojas[i] * media_frete_lojas[i])
    faturamento_menos_despesas.append(lucro)

for i in range(len(nomes_lojas)):
    print(f"{nomes_lojas[i]} - Lucro: R${faturamento_menos_despesas[i]:.2f}")
```

Levando em consideração os valores disponíveis para Analise, como o faturamento, a despesa com o frete, a quantidade de produtos vendidos, presumindo que o lucro obtido na venda de cada produto seja o mesmo e observando a Avaliação da Loja pelos clientes, minha indicação para venda da Loja é a Loja 4. Durante a análise foi possível verificar que a Loja 4 tem o menor faturamento entre as lojas análisadas, e apesar de ter o frete médio menor que as demais lojas, no valor de faturamento final, quando subtraimos o valor do frete do faturamento, continuamos vendo o pior desempenho entre todas as lojas. Também vendo a Avaliação dos Clientes, vemos que ela tem o segundo pior desempenho, ficando apenas acima da Loja 1, que em compensação tem o maior faturamento e faturamento final entre todas as Lojas. Observando todos esses aspectos, nossa conclusão e sugestão é que a Loja 4 seja vendida
